```python
#This script loads images in the folder [image_directory] and runs the images in that folder through the algae classification
#model saved at [model_path]. It outputs the name of each image and the corresponding predicted algae class and probability. The script also
#outputs an excel file with the algae labels to the [results_directory] folder. 
```

```python
# Define the path to the model and the directory containing new images
model_path = 'fill out' #Update to model you want to use (e.g., model_path = '/Documents/algaeproject/models/algaeclassifier_cnn.h5')
image_directory = 'fill out'  # Update this path to your folder of images (e.g., 
results_directory = 'NovelTestBatch2/results' #update to path of folder you want to save excel file in
```

```python
#Import dependencies
import os
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model
import pandas as pd
```

```python
# Load the trained model
model = load_model(model_path)

# Define the class names (as in your training script)
class_names = ['Actinoptychus', 'Bacillaria', 'Biddulphia', 'Centric Diatom', 'Ciliate', 'Coscinodiscus', 'Cylindrotheca',
               'Dactyliosolen', 'Diatom', 'Dinoflagellate', 'Entomoneis', 'Euglenoid', 'Fragilidium', 'Hemiaulus', 'Heterosigma Akashiwo',
               'Lyrella', 'Navicula', 'Nitzschia', 'Odontella', 'Paralia', 'Pennate Diatom', 'Pleurosigma', 'Prorocentrum', 'Tintinnid',
               'Tripos Hircus']

# Function to preprocess the images
def preprocess_image(image_path):
    img = cv2.imread(image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = tf.image.resize(img, (256, 256))
    img = img / 255.0  # Normalize the image
    return np.expand_dims(img, axis=0)  # Expand dimensions to match the model's input forma

# Prepare to accumulate results
results = []

# Iterate over each image in the directory
for filename in os.listdir(image_directory):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp')):  # Check for image formats
        file_path = os.path.join(image_directory, filename)
        image = preprocess_image(file_path)
        prediction = model.predict(image)
        predicted_class_index = np.argmax(prediction)
        predicted_probability = np.max(prediction)
        predicted_class_name = class_names[predicted_class_index]

        # Append results to the list
        results.append({
            "Image": filename,
            "Predicted Class Name": predicted_class_name,
            "Predicted Probability": predicted_probability
        })

results
```
