```python
#print loss and accuracy graphs:


import pandas as pd
import matplotlib.pyplot as plt
#%matplotlib inline


# Load the metrics
results_path = '/Applications/anaconda3/envs/COMP6970/FinalProject/yolov5/runs/train-cls/exp42/results.csv'  # Adjust path accordingly
data = pd.read_csv(results_path)

# Strip whitespace from the column names
data.columns = data.columns.str.strip()

# Now, print the cleaned column names to confirm the change
print(data.columns)

# Print the DataFrame to see available columns
print(data.head())

# Plotting training and validation loss
plt.figure(figsize=(10, 5))
plt.plot(data['train/loss'], label='Training Loss')
plt.plot(data['val/loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()

# Plotting accuracy 
plt.figure(figsize=(10, 5))
plt.plot(data['metrics/accuracy_top1'], label='Top 1 Accuracy')  
plt.plot(data['metrics/accuracy_top5'], label='Top 5 Accuracy')  
plt.title('Top 1 & 5 Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```
